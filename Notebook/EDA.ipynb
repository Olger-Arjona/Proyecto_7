```python
import pandas as pd
import plotly.graph_objects as go 
import plotly.express as px

#lectura de datos
car_data = pd.read_csv('vehicles_us.csv')
car_data.head()
```

```python
car_data.info()
```

```python
# Valores faltantes por columna
car_data.isna().sum()
```

```python
#histograma del odómetro
fig = go.Figure(data=[go.Histogram(x=car_data['odometer'])])
fig.update_layout(title_text='Distribución del Odómetro', xaxis_title='Odómetro (millas)', yaxis_title='Cantidad de anuncios')
fig.show()
```

```python
#histograma del precio
fig = go.Figure(data=[go.Histogram(x=car_data['price'])])
fig.update_layout(title_text='Distribución del Precio', xaxis_title='Precio ($)', yaxis_title='Cantidad de anuncios')
fig.show()
```

```python
#dispersión odómetro vs precio
fig = go.Figure(data=[go.Scatter(x=car_data['odometer'], y=car_data['price'], mode='markers', marker=dict(opacity=0.4, size=4))])
fig.update_layout(title_text='Relación entre Odómetro y Precio', xaxis_title='Odómetro (millas)', yaxis_title='Precio ($)')
fig.show()
```

```python
#precio promedio por tipo de vehículo
avg_price_by_type = car_data.groupby('type')['price'].mean().sort_values(ascending=False)

fig = go.Figure(data=[go.Bar(x=avg_price_by_type.index, y=avg_price_by_type.values)])
fig.update_layout(title_text='Precio Promedio por Tipo de Vehículo', xaxis_title='Tipo', yaxis_title='Precio promedio ($)')
fig.show()
```

```python
#cantidad de anuncios por condición
condition_counts = car_data['condition'].value_counts()

fig = go.Figure(data=[go.Bar(x=condition_counts.index, y=condition_counts.values)])
fig.update_layout(title_text='Cantidad de Anuncios por Condición', xaxis_title='Condición', yaxis_title='Cantidad')
fig.show()
```
